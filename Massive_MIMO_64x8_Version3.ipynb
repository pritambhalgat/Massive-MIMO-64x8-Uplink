# Massive MIMO 64x8 — Version 3: Modular and Reproducible Simulation

Version 3 converts the Version 2 methodology into a reusable Python package. It adds multi-seed evaluation, confidence intervals, unit tests, and receiver-specific post-detection SINR.

The channel model includes large-scale fading, correlated ULA, multipath and AR(1)-approximated Doppler. Orthogonal-pilot LS/LMMSE estimation and MRC/ZF/MMSE receivers are supported. SE/EE remain proxies, not coded goodput or calibrated energy measurements.

```python
from pathlib import Path
import os, sys, subprocess
ROOT = Path.cwd()
V3 = ROOT / 'version3'
if not V3.exists(): raise FileNotFoundError('Open notebook from repository root.')
os.chdir(V3); sys.path.insert(0, str(V3))
print('Version 3 path:', V3)
```

## Install and test

Run unit tests before the Monte Carlo experiment.

```python
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'requirements.txt'], check=True)
subprocess.run([sys.executable, '-m', 'pytest', '-q'], check=True)
```
